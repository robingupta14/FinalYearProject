```python
import os
DATASET_ROOT = "../../CrossVul"
NEW_DATASET = "../../Preprocessed"
CWE_IDS = {"CWE-22"} # "CWE-22", "CWE-89", "CWE-787"
LANGUAGES = ['c', 'cpp', 'cs', 'html', 'java', 'py', 'php']
```

```python
def preprocess(code, lang):
    match lang:
        case 'c':
            return code
        case 'cpp':
            return code
        case 'cs':
            return code
        case 'html':
            return code
        case 'java':
            return code
        case 'py':
            return code
        case 'php':
            return code
        case _:
            return code
```

# Final File Writing

```python
def collect_files_for_cwe(cwe_id):
    for lang in LANGUAGES:
        lang_dir = os.path.join(DATASET_ROOT, cwe_id, lang)
        if not os.path.isdir(lang_dir):
            continue
        for filename in os.listdir(lang_dir):
            filepath = os.path.join(lang_dir, filename)
            if filename.endswith('.DS_Store'):
                continue
            with open(filepath, 'r', encoding='utf-8', errors='ignore') as f:
                code = f.read()
                
            out_dir = os.path.join(NEW_DATASET, cwe_id, lang)
            os.makedirs(out_dir, exist_ok=True)
            out_path = os.path.join(out_dir, filename)
            preprocessed_code = preprocess(code, lang)

            with open(out_path, 'w', encoding='utf-8') as out_f:
                out_f.write(preprocessed_code)
```
